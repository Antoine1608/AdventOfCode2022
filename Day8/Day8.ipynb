```python
from itertools import product
```

```python
with open(r"C:\Users\John\Desktop\AdventOfCode\AdventOfCode2022\Day8\input.txt") as f:
    input = f.read()\
             .split('\n')\
             [:-1]
input = [list(map(lambda x : int(x), input[i])) for i in range(len(input))]
```

```python
# Part 1
count = 0
for i,j in product(range(len(input)), range(len(input[0]))):
    if all([input[i][j]> x for x in input[i][j+1:]]):
        #print('vue de Droite : ', (i, j))
        count += 1
    elif all([input[i][j]> x for x in input[i][:j]]):
        #print('vue de Gauche : ', (i, j))
        count += 1
    elif all([input[i][j]> x for x in [r[j] for r in input[i+1:]]]):
        #print('vue du bas : ', (i, j))
        count += 1
    elif all([input[i][j]> x for x in [r[j] for r in input[:i]]]):
        #print('vue du haut : ', (i, j))
        count += 1
count
```

```python
# part 2
# Rm : 3 1 2 1 => le 3 voit 3 arbres et non 2 !
score = 0

def view(caban : tuple, dire : str) -> int:
    i, j = caban
    #print('caban : ', caban)
    H = input[i][j]
    n = 0
    prev = 0
    if dire == 'right':
        liste = input[i][j+1:]#j+1
    if dire == 'left':
        liste = input[i][:j][::-1]
    if dire =='up':
        liste = [r[j] for r in input[:i]][::-1]
    if dire == 'down':
        liste = [r[j] for r in input[i+1:]]#i+1

    #print('liste : ', liste)
    #print('H : ', H)
    
    for tree in liste:
        n += 1
        #print('tree : ', tree)
        if tree >= H:
            break

    return n

view((3, 2), 'up')
```

```python
score = 0
for i,j in product(range(len(input)), range(len(input[0]))):
    score_tree = view((i, j), 'up') * view((i, j), 'down') * view((i, j), 'right') * view((i, j), 'left')
    if score_tree > score:
        score = score_tree
        place = (i, j)

print("score : ", score, " pour : ", place)
#3240 too low
```
