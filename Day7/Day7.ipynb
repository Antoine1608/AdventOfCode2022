```python
import numpy as np
from bigtree import Node, print_tree
import re
from collections import deque
```

```python
with open(r"C:\Users\John\Desktop\AdventOfCode\AdventOfCode2022\Day7\input.txt") as f:
    lines = f.read().splitlines()
lines
```

```python
tree = {0: Node(name='/', size=0, type='dir')}
current = tree[0]
for i, line in enumerate(lines):
    if '$ cd' in line:
        loc = line.split(' ')[-1]
        if loc == '/':
            current = tree[0]
        elif loc == '..':
            current = current.parent
        else:
            current = next(c for c in current.children if c.name == loc)
    elif '$ ls' in line:
        j = i + 1
        while j < len(lines) and '$' not in lines[j]:
            if 'dir' in lines[j]:
                tree[j] = Node(name=lines[j].split(' ')[-1], size=0, type='dir', parent=current)
            else:
                tree[j] = Node(name=lines[j].split(' ')[-1],
                               size=int(re.findall(r'\d+', lines[j])[0]),
                               type='file', parent=current)
            j += 1
```

```python
print_tree(tree[0], attr_list=["size", "type"])
```

```python
def get_size(node: Node):
    if node.type == 'dir':
        node.size = sum(get_size(child) for child in node.children)
    return node.size

get_size(tree[0])
```
